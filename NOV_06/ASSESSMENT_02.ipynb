```sql
%python
from pyspark.sql.functions import col, when

category_df = registrations_df.withColumn(
    "total_amount",
    col("tickets") * col("unit_price")
).withColumn(
    "booking_category",
    when(col("total_amount") >= 2000, "High")
    .otherwise("Regular")
)

display(category_df)
```

```sql
%sql
SELECT i.registration_id
FROM incoming_registrations i
INNER JOIN event_registrations_work w
ON i.registration_id = w.registration_id;
```

```sql
%sql
SELECT i.registration_id
FROM incoming_registrations i
LEFT ANTI JOIN event_registrations_work w
ON i.registration_id = w.registration_id;
```

```sql
%sql
MERGE INTO event_registrations_work AS target
USING incoming_registrations AS source
ON target.registration_id = source.registration_id

WHEN MATCHED THEN
UPDATE SET
    target.event_id = source.event_id,
    target.attendee_name = source.attendee_name,
    target.registration_date = source.registration_date,
    target.ticket_type = source.ticket_type,
    target.tickets = source.tickets,
    target.unit_price = source.unit_price,
    target.status = source.status,
    target.channel = source.channel

WHEN NOT MATCHED THEN
INSERT (
    registration_id,
    event_id,
    attendee_name,
    registration_date,
    ticket_type,
    tickets,
    unit_price,
    status,
    channel
)
VALUES (
    source.registration_id,
    source.event_id,
    source.attendee_name,
    source.registration_date,
    source.ticket_type,
    source.tickets,
    source.unit_price,
    source.status,
    source.channel
);
```

```sql
%python
from pyspark.sql.functions import col, concat, lit

transactions_df = (
    rate_df
    .withColumn(
        "transaction_id",
        concat(lit("TXN-"), col("value").cast("string"))
    )
    .withColumn(
        "transaction_amount",
        col("value") * 100
    )
)

display(transactions_df)
```

```sql
%python
high_value_df = transactions_df.filter(
    col("transaction_amount") > 500
)

display(high_value_df)
```

```sql
%python
from pyspark.sql.functions import when, count

categorized_df = transactions_df.withColumn(
    "transaction_category",
    when(col("transaction_amount") >= 1000, "High")
    .otherwise("Regular")
)
```
