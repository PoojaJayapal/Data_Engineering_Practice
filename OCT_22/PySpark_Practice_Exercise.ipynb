**Exercise Set 4** — withColumn()

**Create:gross_amount
using:
quantity × unit_price**

```python
df2=df.withColumn("gross_amount",col("quantity")*col("unit_price"))
df2.show()
```

**23. Add the gross_amount column.**

```python
df2=df.withColumn("gross_amount",col("quantity")*col("unit_price"))
df2.show()
```

**24. Create:
discount_amount
using:
gross_amount × discount_pct / 100**

```python
df3 = df2.withColumn("discount_amount", col("gross_amount") * col("discount_pct") / 100)
df3.show()
```

**25. Create:
final_amount
using:
gross_amount - discount_amount**

```python
df3=df3.withColumn("final_amount",col("gross_amount")-col("discount_amount"))
df3.show()
```

**26. Add a constant column:
country = India**

```python
from pyspark.sql.functions import lit
df3=df3.withColumn("country",lit("India"))
df3.show()
```

**27. Add:
high_value
where:
final_amount >= 50000 → Yes
otherwise → No**

```python
from pyspark.sql.functions import when
df3=df3.withColumn("high_value",when(col("final_amount")>=5000,"Yes").otherwise("No"))
df3.show()
```

**Exercise Set 5** — Conditional Columns

**Create a column:
order_size
using:
final_amount >= 50000 → High
final_amount >= 15000 → Medium
otherwise → Low**

```python
from pyspark.sql.functions import col, when

df3 = df3.withColumn(
    "order_size",
    when(col("final_amount") >= 15000, "High")
    .when(col("final_amount") >= 5000, "Medium")
    .otherwise("Low")
)
df3.select("order_id", "final_amount", "order_size").show()
```

**29. Create another column:
quantity_type
where:
quantity >= 4 → Bulk
otherwise → Regular**

```python
df3=df3.withColumn("quantity_type",when (col("quantity")>=4,"Bulk").otherwise("Regular"))
df3.show()
```

**30. Create a delivery-status category where:**

```python
df3 = df3.withColumn(
    "delivery_status",
    when(col("status") == "Delivered", "Completed")
    .when(col("status").isin("Cancelled", "Returned"), "Failed")
    .otherwise("In Progress")
)
df3.select("order_id", "status", "delivery_status").show()
```
